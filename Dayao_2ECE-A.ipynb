# A. Word rotation problem

## Base code

```python
def rotate_word(x):    #name of the function
    y = x[:1] #cuts and saves the first letter of the input variable into variable y by removing every letter aside from the first
    x = x[1:] #saves every letter of the input variable except for the first into variable y
    x = (x+y) #combines the strings together
    return x

rotate_word("test")
```

# B. Username builder problem

## Base Code

```python
def make_username(x,y):
    x = x.lower().replace(" ","") #applies lowercase to characters and removes spaces for first names
    y = y.lower().replace(" ","") #applies lowercase to characters and removes spaces for last names
    x = (x)+(".")+(y) #joints the 2 names together with period in the middle
    return (x)

make_username("test", "1")
```

# C. Bookend Swap Problem

## Base Code

```python
def swap_bookends(x):
    first, *middle, last = x #assigns each variable to their respective list element 
    middle.append(first) #adds the first element to the end of the list
    middle.insert(0, last) #adds the last element to the start of the list
    return middle

swap_bookends(["Test", "1"])
```
